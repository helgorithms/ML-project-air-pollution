```python
# import libraries

import pandas as pd
import numpy as np
import seaborn as sns

import matplotlib.pyplot as plt
```

```python
# load data set

df_test = pd.read_csv("data/Train.csv")
```

```python
# heatmap of missing values

# Missing values matrix (0=complete, 1=missing)
missing_matrix = df_test.isnull()

plt.figure(figsize=(12, 8))
sns.heatmap(missing_matrix, 
            #cbar=True, 
            yticklabels=False, 
            cmap='viridis')

plt.xlabel('Features')
plt.ylabel('Rows')
plt.tight_layout()

# first save then show
plt.savefig('images/plot_missing_values.png', dpi=300, bbox_inches='tight')

plt.show()
```
